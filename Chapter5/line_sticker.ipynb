```python
import requests
from bs4 import BeautifulSoup as bs

url = "https://store.line.me/stickershop/product/17985995/zh-Hant"

r = requests.get(url)

print(r)

soup = bs(r.text, 'lxml')

# print(soup)
```

```python
import json
import os

folder = 'line_stickers'
if not os.path.exists(folder):
    os.mkdir('line_stickers')

elements = soup.select("ul.mdCMN09Ul.FnStickerList > li")
print(len(elements))

n = 0
for element in elements:
    print(element)
    img_element = json.loads(element['data-preview'])
    img_url = img_element['staticUrl']
    print(img_url)
    n += 1

    with open (f'{folder}/stickers_{n}.png', mode='wb') as img:
        img.write(requests.get(img_url).content)
```
