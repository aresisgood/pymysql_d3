建議的寫法

```python
connection = pymysql.connect(
    host='localhost',
    user='root', 
    password='P@ssw0rd',
    port=3306, 
)

# 建立游標
with connection.cursor() as cursor:
    # 使用游標執行 SQL 語句
    sql = """
        SHOW DATABASES
    """
    cursor.execute(sql)
    
    # 取得所有資料
    result = cursor.fetchall()
    print(result)
```

```python
from pprint import pprint

# 使用 DB > TABLE 查詢
with connection.cursor() as cursor:
    # 使用游標執行 SQL 語句
    sql = """
        SHOW TABLES FROM my_titanic;
    """
    cursor.execute(sql)
    
    # 取得所有資料
    result = cursor.fetchall()
    print(result)

    sql = """
        SELECT * FROM my_titanic.full_passengers LIMIT 10;
    """
    cursor.execute(sql)

    # 取得所有資料
    result = cursor.fetchall()
    print(result)
```

指定資料庫

```python
import pymysql
connection = pymysql.connect(
    host='localhost',
    user='root', 
    password='P@ssw0rd',
    port=3306,
    # 連線時指定資料庫名稱
    database='my_titanic'
)

# 建立游標
with connection.cursor() as cursor:
    # 使用游標執行 SQL 語句
    sql = "SHOW TABLES;"
    cursor.execute(sql)
    # 取得所有資料
    databases = cursor.fetchall()

print(databases)
```

使用 cursorclass 修改回傳資料格式

```python
# 資料庫連線
connection = pymysql.connect(
    host='localhost',
    user='root', 
    password='P@ssw0rd',
    port=3306,
    database="my_titanic",
    # 使用字典型游標
    cursorclass=pymysql.cursors.DictCursor
)

# 建立游標
with connection.cursor() as cursor:
    # 使用游標執行 SQL 語句
    sql = "SHOW DATABASES;"
    cursor.execute(sql)
    # 取得所有資料
    databases = cursor.fetchall()

print(databases)
```

查詢資料庫

```python
from pprint import pprint

with connection.cursor() as cursor:
    # 使用游標執行 SQL 語句
    sql = "SELECT * FROM full_passengers LIMIT 3;"
    cursor.execute(sql)
    # 取得所有資料
    result = cursor.fetchall()

print(result)
pprint(result[0], sort_dicts=False)
```
