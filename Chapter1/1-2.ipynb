使用設定檔建立資料庫連線

```python
import pymysql
from configparser import ConfigParser

# 讀取 config.ini 檔案取得資料庫連線資訊
config = ConfigParser()
config.read('config.ini')

# 讀取 config.ini 檔案取得資料庫連線資訊

# 建立資料庫連線
connection = pymysql.connect(
    host=config['DB']['host'],
    user=config['DB']['user'],
    password=config['DB']['password'],
    port=int(config['DB']['port']), # 轉成 int
    cursorclass=pymysql.cursors.DictCursor,
)

with connection.cursor() as cursor:
    query = """
        SHOW DATABASES;
    """
    cursor.execute(query)
    result = cursor.fetchall()
print(result)
```

```python
print(config.get('DB', 'host'))
print(config.get('DB', 'password'))
print(config.getint('DB', 'port')) # 拿出時直接轉成 int
```

使用 SQL 函式

```python
def sql_query(sql):
    with connection.cursor() as cursor:
        cursor.execute(sql)
        result = cursor.fetchall()

    return result

data = sql_query("SHOW DATABASES;")
print(data)
```
